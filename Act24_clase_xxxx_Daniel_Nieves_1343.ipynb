```python
class Parque:
    def __init__(self, nombre, extensionHectareas, tipoVegetacion, atractivos):
        self.nombre = nombre
        self.extensionHectareas = extensionHectareas
        self.tipoVegetacion = tipoVegetacion
        self.atractivos = atractivos

    def mantenerAreasVerdes(self):
        print(f"\n >>> [MANTENIMIENTO]: Iniciando poda y riego en el parque {self.nombre}.")
        print(f" >>> Cuidando la vegetación de tipo: {self.tipoVegetacion}.")

    def organizarActividad(self):
        actividad = input("¿Qué actividad desea organizar? (Ej: Yoga, Concierto, Picnic): ")
        print(f"\n >>> [EVENTO]: Organizando '{actividad}' en el área de {self.atractivos}.")

    def vigilarSeguridad(self):
        print(f"\n >>> [SEGURIDAD]: Desplegando guardaparques en las {self.extensionHectareas} hectáreas del parque.")

# --- Inicio del Programa Interactivo ---

print("="*40)
print("SISTEMA DE GESTIÓN DE PARQUES")
print("="*40)

# Captura de datos del usuario
nombre_p = input("Ingrese el nombre del Parque: ")
ext_p = input("Ingrese la extensión en hectáreas: ")
veg_p = input("Ingrese el tipo de vegetación dominante: ")
atr_p = input("Ingrese el atractivo principal: ")

# Creación del objeto con los datos ingresados
parque_usuario = Parque(nombre_p, ext_p, veg_p, atr_p)

print(f"\n--- Parque '{parque_usuario.nombre}' registrado con éxito ---")

# Menú de acciones para interactuar
while True:
    print("\n¿Qué acción de gestión desea realizar?")
    print("1. Mantener áreas verdes")
    print("2. Organizar actividad")
    print("3. Vigilar seguridad")
    print("4. Ver detalles del parque")
    print("5. Salir")

    opcion = input("\nSeleccione una opción (1-5): ")

    if opcion == "1":
        parque_usuario.mantenerAreasVerdes()
    elif opcion == "2":
        parque_usuario.organizarActividad()
    elif opcion == "3":
        parque_usuario.vigilarSeguridad()
    elif opcion == "4":
        print(f"\n--- FICHA TÉCNICA ---")
        print(f"Nombre: {parque_usuario.nombre}")
        print(f"Extensión: {parque_usuario.extensionHectareas} ha")
        print(f"Vegetación: {parque_usuario.tipoVegetacion}")
        print(f"Atractivos: {parque_usuario.atractivos}")
    elif opcion == "5":
        print("\nCerrando sistema de gestión ambiental...")
        break
    else:
        print("\n[!] Opción no válida, intente de nuevo.")

print("\nFin del programa")
print("Programa creado por Daniel Nieves NL=41 NC=1343")
```
